```python
#ej1
def contar_frecuencia_letras(texto):
    """
    Recibe una cadena de texto y devuelve un diccionario
    con la frecuencia de cada letra (los espacios no se cuentan).
    """
    frecuencias = {}  # Diccionario vacío donde iremos acumulando resultados

    for caracter in texto:
        if caracter == " ":
            continue  # Saltamos los espacios, no queremos contarlos

        # .get(caracter, 0) es la clave del ejercicio:
        # busca 'caracter' en el diccionario y, si no existe todavía,
        # devuelve 0 en lugar de dar error. Así podemos sumar 1 directamente
        # sin tener que comprobar antes con un if si la clave existe.
        frecuencias[caracter] = frecuencias.get(caracter, 0) + 1

    return frecuencias


# Ejemplo de uso
if __name__ == "__main__":
    resultado = contar_frecuencia_letras("hola mundo")
    print(resultado)
```

```python
#ej2
def duplicar_valores(numeros):
    """
    Recibe una lista de números y devuelve una nueva lista
    con cada valor multiplicado por 2, usando map().
    """
    # map() aplica la función lambda a cada elemento de 'numeros'
    # y devuelve un objeto 'map' (un iterador), no una lista directamente.
    # Por eso hay que envolverlo en list() para obtener la lista final.
    return list(map(lambda x: x * 2, numeros))


# Ejemplo de uso
if __name__ == "__main__":
    numeros = [1, 2, 3, 4, 5]
    resultado = duplicar_valores(numeros)
    print(resultado)
```

```python
#ej3
def filtrar_palabras_que_contienen(lista_palabras, palabra_objetivo):
    """
    Recibe una lista de palabras y una palabra objetivo.
    Devuelve una lista con las palabras de la lista original
    que contienen la palabra objetivo como subcadena.
    """
    # Usamos una list comprehension con un condicional dentro:
    # recorremos cada 'palabra' y la incluimos en la lista resultante
    # solo si 'palabra_objetivo' está contenida en ella.
    # El operador 'in' en cadenas de texto comprueba subcadenas,
    # no solo coincidencias exactas de elemento.
    return [palabra for palabra in lista_palabras if palabra_objetivo in palabra]


# Ejemplo de uso
if __name__ == "__main__":
    palabras = ["casa", "casita", "perro", "casona", "gato"]
    objetivo = "casa"
    resultado = filtrar_palabras_que_contienen(palabras, objetivo)
    print(resultado)
```

```python
#ej4
def diferencia_listas(lista1, lista2):
    """
    Recibe dos listas de números y devuelve una nueva lista
    con la diferencia elemento a elemento (lista1[i] - lista2[i]),
    usando map().
    """
    # A diferencia del ejercicio 2, aquí map() recibe DOS iterables.
    # Cuando se le pasan varios iterables, la función lambda recibe
    # un argumento por cada uno, y map() los recorre en paralelo
    # (emparejando el elemento i de lista1 con el elemento i de lista2).
    return list(map(lambda a, b: a - b, lista1, lista2))


# Ejemplo de uso
if __name__ == "__main__":
    lista1 = [10, 20, 30, 40]
    lista2 = [1, 2, 3, 4]
    resultado = diferencia_listas(lista1, lista2)
    print(resultado)
```

```python
#ej 5
def evaluar_media(numeros, nota_aprobado=5):
    """
    Recibe una lista de números y un valor opcional nota_aprobado (por defecto 5).
    Calcula la media de los números y determina si está aprobado o suspenso.
    Devuelve una tupla (media, estado).
    """
    # sum() suma todos los elementos, len() cuenta cuántos hay.
    # Dividir ambos nos da la media aritmética.
    media = sum(numeros) / len(numeros)

    # Comparamos la media con el umbral de aprobado.
    # Como nota_aprobado es un parámetro con valor por defecto,
    # si no se pasa al llamar a la función, se usará 5 automáticamente.
    estado = "aprobado" if media >= nota_aprobado else "suspenso"

    # Devolvemos una tupla: en Python basta con separar los valores
    # por comas (los paréntesis son opcionales pero mejoran la legibilidad).
    return (media, estado)


# Ejemplo de uso
if __name__ == "__main__":
    notas = [4, 6, 5, 7]
    resultado = evaluar_media(notas)
    print(resultado)

    resultado2 = evaluar_media(notas, nota_aprobado=6)
    print(resultado2)
```

```python
#ej6
def factorial(n):
    """
    Calcula el factorial de un número entero no negativo n
    de forma recursiva (n! = n * (n-1) * (n-2) * ... * 1).
    """
    # Caso base: es la condición que detiene la recursión.
    # Sin un caso base, la función se llamaría a sí misma
    # infinitamente hasta provocar un error (RecursionError).
    # Por definición, 0! = 1 y 1! = 1.
    if n == 0 or n == 1:
        return 1

    # Caso recursivo: la función se llama a sí misma con un
    # problema más pequeño (n-1) hasta llegar al caso base.
    # Por ejemplo: factorial(4) = 4 * factorial(3)
    #                            = 4 * 3 * factorial(2)
    #                            = 4 * 3 * 2 * factorial(1)
    #                            = 4 * 3 * 2 * 1
    return n * factorial(n - 1)


# Ejemplo de uso
if __name__ == "__main__":
    print(factorial(5))
    print(factorial(0))
```

```python
#ej7
def tuplas_a_strings(lista_tuplas):
    """
    Recibe una lista de tuplas y devuelve una lista de strings,
    donde cada tupla se convierte en un string con sus elementos
    separados por comas, usando map().
    """
    # Función auxiliar que convierte UNA tupla en un string.
    # Usamos join() en lugar de concatenar con '+' porque es más
    # eficiente y es la forma "pythónica" de unir varios valores en uno.
    def tupla_a_string(tupla):
        # join() solo funciona con strings, así que primero convertimos
        # cada elemento de la tupla a str() con una comprehension,
        # por si la tupla contiene números u otros tipos de datos.
        return ", ".join(str(elemento) for elemento in tupla)

    return list(map(tupla_a_string, lista_tuplas))


# Ejemplo de uso
if __name__ == "__main__":
    tuplas = [(1, "manzana"), (2, "pera"), (3, "uva")]
    resultado = tuplas_a_strings(tuplas)
    print(resultado)
```

```python
#ej8
def dividir_numeros():
    """
    Pide al usuario dos números y los divide, manejando los posibles
    errores: valor no numérico o división por cero.
    """
    try:
        # input() siempre devuelve un string, así que hay que convertirlo
        # a float explícitamente. Esta conversión es la que puede fallar
        # si el usuario escribe algo que no es un número (p. ej. "hola").
        numero1 = float(input("Introduce el primer número: "))
        numero2 = float(input("Introduce el segundo número: "))

        resultado = numero1 / numero2

    except ValueError:
        # Salta si float() no puede convertir el texto introducido.
        print("Error: debes introducir valores numéricos.")

    except ZeroDivisionError:
        # Salta específicamente si numero2 es 0.
        # Es importante capturar esta excepción ANTES o por separado
        # de una excepción genérica, porque es más específica.
        print("Error: no se puede dividir entre cero.")

    else:
        # El bloque 'else' de un try solo se ejecuta si NO ha saltado
        # ninguna excepción. Es una buena práctica separarlo del 'try'
        # para distinguir claramente "código que puede fallar" de
        # "código que depende de que todo haya ido bien".
        print(f"División exitosa: {numero1} / {numero2} = {resultado}")

    finally:
        # Se ejecuta siempre, haya habido error o no.
        # Útil para mensajes de cierre o liberar recursos.
        print("Programa finalizado.")


# Ejecución del programa
if __name__ == "__main__":
    dividir_numeros()
```

```python
#ej9
# Definimos la lista de mascotas prohibidas fuera de la función,
# como una constante a nivel de módulo (en mayúsculas, por convención,
# ya que es un valor fijo que no cambia durante la ejecución).
MASCOTAS_PROHIBIDAS = ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]


def filtrar_mascotas_permitidas(lista_mascotas):
    """
    Recibe una lista de nombres de mascotas y devuelve una nueva lista
    excluyendo las que están prohibidas en España, usando filter().
    """
    # filter(funcion, iterable) recorre 'lista_mascotas' y se queda
    # solo con los elementos para los que la función devuelve True.
    # A diferencia de map() (que TRANSFORMA cada elemento), filter()
    # SELECCIONA elementos sin modificarlos.
    return list(filter(lambda mascota: mascota not in MASCOTAS_PROHIBIDAS, lista_mascotas))


# Ejemplo de uso
if __name__ == "__main__":
    mascotas = ["Perro", "Gato", "Mapache", "Tigre", "Hámster", "Oso"]
    resultado = filtrar_mascotas_permitidas(mascotas)
    print(resultado)
```

```python
#ej10
class ListaVaciaError(Exception):
    """
    Excepción personalizada que se lanza cuando se intenta
    calcular el promedio de una lista vacía.

    Heredar de Exception (la clase base de todas las excepciones
    en Python) es lo que permite que Python la trate como un error
    real y que pueda capturarse con 'except ListaVaciaError'.
    """
    pass


def calcular_promedio(numeros):
    """
    Recibe una lista de números y devuelve su promedio.
    Lanza ListaVaciaError si la lista está vacía.
    """
    # 'not numeros' es True si la lista está vacía (las listas vacías
    # son "falsy" en Python). Es más idiomático que comparar con
    # 'len(numeros) == 0'.
    if not numeros:
        # 'raise' lanza la excepción de forma manual. Le pasamos un
        # mensaje descriptivo que podrá leerse después al capturarla.
        raise ListaVaciaError("No se puede calcular el promedio de una lista vacía.")

    return sum(numeros) / len(numeros)


# Ejecución del programa, manejando la excepción personalizada
if __name__ == "__main__":
    listas_de_prueba = [[10, 20, 30], []]

    for lista in listas_de_prueba:
        try:
            promedio = calcular_promedio(lista)
        except ListaVaciaError as error:
            # 'as error' guarda el objeto de la excepción, y str(error)
            # (o simplemente imprimirlo) da acceso al mensaje que le
            # pasamos al hacer 'raise'.
            print(f"Error: {error}")
        else:
            print(f"El promedio es: {promedio}")
```
